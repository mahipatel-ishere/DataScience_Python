# Box plot to visulize the outlier


- Boxplot is an effective statistical too to visulaize the outlier and other metrics


````python
plt.boxplot(df["Marks"])
plt.title("Boxplot Of Marks Column")
plt.show()
````

# Function to detect the outlier



````python
def detectoutlier(df,column):
    q1=df[column].quantile(0.25)
    q3=df[column].quantile(0.75) 
    
    iqr=q3-q1
    
    lower=q1-1.5*iqr
    upper=q3+1.5*iqr    
    
    outlier=df[(df[column]<lower) | (df[column]>upper) ]
    return outlier
````

````python
detectoutlier(df,'Marks')
````

# Function to Remove Outlier

````python
def removeoutlier(df,column):
    q1=df[column].quantile(0.25)
    q3=df[column].quantile(0.75) 
    
    iqr=q3-q1
    
    lower=q1-1.5*iqr
    upper=q3+1.5*iqr    
    
    cleandf=df[(df[column]>=lower) & (df[column]<=upper) ]
    return cleandf
````

````python
removeoutlier(df,'Marks')
````

````python
df1=removeoutlier(df,'Marks')
````

````python
plt.boxplot(df1['Marks'])
plt.show()
````
